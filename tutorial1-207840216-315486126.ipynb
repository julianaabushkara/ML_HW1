Write a function named **cube_volume** that when given a length of a line, returns the area of a cube.

For exmaple cube_volume(7) returns 343.

```python
def cube_volume(len):
    if(len<=0):
        print('ERROR please enter a positive number!')
        return
    return (len**3)
    
#we assumed that you can input any length and it prints an error if <0
cube_len = int(input('Please enter a cube len'))
print(f'the cube volume : {cube_volume(cube_len)}')

#print(f'the cube volume : {cube_volume(7)}')
```

## Assignment #2

Write a function named **parse_words**, that given a string prints each word in a different line.

<p> For example parse_words ("<em><strong>Lorem ipsum dolor sit amet, vis dolore soleat</em></strong>")  will pe printed as</p>
<p> Lorem </p>
<p> ipsum </p>
<p> dolor </p>
<p> sit </p>
<p> amet, </p>
<p> vis </p>
<p> dolore </p>
<p> soleat </p>

```python
def parse_words(lines):
    if lines:
        names_list = lines.split()
        for s in names_list:
            print(f'{s} \n')
    else :
        print("string is empty")
    return

str_line = input('Please enter a string:')

print('\n')
parse_words(str_line)

```

<p> Write a function <strong> today_date </strong> that prints today's date in the format DD-MM-YYYY. </p>
<p> For example if today is March 10, 2023, the function will print <em> 10-03-2023 </em>. </p>

```python
from datetime import datetime

def today_date():
    current_date = datetime.now()
    Day= str(current_date.day).zfill(2)
    Month =str(current_date.month).zfill(2)
    Year=str(current_date.year)
    print(f'Today is: {Day}-{Month}-{Year}')
    return

today_date()

```
